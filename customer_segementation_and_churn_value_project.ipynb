Approach ocombonation of trees and depth

```python
from sklearn.metrics import accuracy_score,recall_score,precision_score,f1_score
```

```python
n_estimators_list=[100,200,300,400,500]
max_depth_list=[5,10,15,20]
result=[]
for n_tress in n_estimators_list:
  for depth in max_depth_list:
    rf_model=RandomForestClassifier(n_estimators=n_tress,max_depth=depth,random_state=42,class_weight='balanced')
    rf_model.fit(X_train,Y_train)
    y_pred=rf_model.predict(X_test)
    accuracy=accuracy_score(Y_test,y_pred)
    recall=recall_score(Y_test,y_pred)
    precision=precision_score(Y_test,y_pred)
    f1=f1_score(Y_test,y_pred)
    result.append({'Trees':n_tress,'depth':depth,'accuracy':accuracy,'recall':recall,'precision':precision,'f1 Score':f1})
    result_df=pd.DataFrame(result)
    result_df=result_df.sort_values(by=['recall', 'accuracy'],ascending=False)
    print(result_df.head(20))
```

**Customer Segementation**

```python
y_prob_full=rf_tuned.predict_proba(X)
churn_prob_full=y_prob_full[:,1]
segmentation_data=pd.DataFrame({
    'Tenure Months':X['Tenure Months'],
    'Monthly Charges':X['Monthly Charges'],
    'Total Charges':X['Total Charges'],
    'Churn Probablity': churn_prob_full
})
```

```python
display(segmentation_data)
```

**implementation of kmeans**

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

```

```python
scaled_data = scaler.fit_transform(segmentation_data)
print(scaled_data[:5])
```

```python
from sklearn.cluster import KMeans

WCSS=[]
for k in range(1,16):
  kmeans=KMeans(n_clusters=k,random_state=42, n_init='auto')
  kmeans.fit(scaled_data)
  WCSS.append(kmeans.inertia_)

plt.figure(figsize=(8,6))
plt.plot(range(1,16),WCSS, marker='o')
plt.title('Elbow Method For Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS (Within-Cluster Sum of Squares)')
plt.xticks(range(1, 16))
plt.grid(True)
plt.show()
```

```python
kmeans=KMeans(n_clusters=3,random_state=42)
```

```python
clusters=kmeans.fit_predict(scaled_data)
```

```python
segmentation_data
```

```python
segmentation_data['Cluster']=clusters
```

```python
segmentation_data
```

```python
cluster_summary=segmentation_data.groupby('Cluster').mean()
```

```python
cluster_summary
```

```python
cluster_summary={
    0:'Budget  Loyal Customers',
    1:'High Risk New Customers',
    2:'Loyal Premium Customers'
}
```

```python
segmentation_data['Cluster Segment ']=segmentation_data['Cluster'].map(cluster_summary)
```
